```python
question_mark_counts = (df == '?').sum()

print(question_mark_counts[
    question_mark_counts > 0
].sort_values(ascending=False))
```

```python
missing_summary = pd.DataFrame({
    'Missing_Count': (df == '?').sum(),
    'Missing_Percentage': (df == '?').mean() * 100
})

missing_summary = missing_summary[
    missing_summary['Missing_Count'] > 0
].sort_values(
    by='Missing_Percentage',
    ascending=False
)

display(missing_summary)
```

```python
id_columns = [
    'admission_type_id',
    'discharge_disposition_id',
    'admission_source_id'
]

for col in id_columns:
    print("\n" + "=" * 60)
    print(f"COLUMN: {col}")
    print("=" * 60)

    print(df[col].value_counts(dropna=False).sort_index())
```

```python
# Select the admission_type_id column
print(df['admission_type_id'])

# Count each ID
df['admission_type_id'].value_counts()
```

```python
# Check actual missing values recognised by Pandas

print("Actual missing values in each column:")
print(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
columns_to_check = [
    'weight',
    'medical_specialty',
    'payer_code',
    'race',
    'diag_1',
    'diag_2',
    'diag_3'
]

for col in columns_to_check:

    print("\n" + "=" * 60)
    print(f"COLUMN: {col}")
    print("=" * 60)

    missing_count = (df[col] == '?').sum()
    missing_percentage = (df[col] == '?').mean() * 100

    print("Missing count:", missing_count)
    print("Missing percentage:", round(missing_percentage, 2), "%")

    print("\nMost common available values:")

    available_values = df.loc[df[col] != '?', col]
    
    print(available_values.value_counts().head(10))
```
