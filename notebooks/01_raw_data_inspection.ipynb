```python

# Create a working copy so the original raw dataset remains unchanged
df_work = df.copy()

# Replace question mark markers with proper missing values
df_work = df_work.replace("?", pd.NA)

# Calculate missing-value counts and percentages
missing_summary = pd.DataFrame({
    "Missing_Count": df_work.isna().sum(),
    "Missing_Percentage": (df_work.isna().mean() * 100).round(2)
})

# Exclude weight from our analysis
missing_summary = missing_summary.drop(index="weight", errors="ignore")

# Display columns that contain missing values
missing_summary = (
    missing_summary[missing_summary["Missing_Count"] > 0]
    .sort_values(by="Missing_Percentage", ascending=False)
)

display(missing_summary)
```

```python
# Columns where missing values need a consistent category
missing_columns = [
    "max_glu_serum",
    "A1Cresult",
    "medical_specialty",
    "payer_code",
    "race",
    "diag_3",
    "diag_2",
    "diag_1"
]

# Replace missing values with a clear label
df_work[missing_columns] = df_work[missing_columns].fillna("Not Recorded")

# Check that these columns no longer contain missing values
print("Remaining missing values:")
display(df_work[missing_columns].isna().sum())

print("Updated dataset shape:", df_work.shape)
```

## 4. Feature Engineering

```python
# 1. Total previous year heathcare visits

visit_columns = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient"
]

for col in visit_columns:
    df_work[col] = pd.to_numeric(df_work[col], errors="coerce")

df_work["total_prior_visits"] = df_work[visit_columns].sum(axis=1)

# 2. Count diabetes-related medications prescribed

medication_columns = [
    "metformin", "repaglinide", "nateglinide", "chlorpropamide", "glimepiride",
    "acetohexamide", "glipizide", "glyburide", "tolbutamide", "pioglitazone", 
    "rosiglitazone", "acarbose", "miglitol", "troglitazone", "tolazamide",
    "examide", "citoglipton", "insulin", "glyburide_metformin", "glipizide_metformin",
    "glimepiride_pioglitazone", "metformin_rosiglitazone", "metformin_pioglitazone"
]

#Use medication columns that exist in the dataset 
available_medications = [
    col for col in medication_columns if col in df_work.columns
]

df_work["medication_count"] = (
    df_work[available_medications]
    .apply(lambda col: col.isin(["steady", "up", "Down"]).astype(int))
    .sum(axis=1)
)

# 3. Medication change indicator

df_work["medication_change_flag"] = (
    df_work["change"].eq("Ch").astype(int)
)

# 4. Diabetes medication indicator

df_work["diabetes_medication_flag"] = (
    df_work["diabetesMed"].eq("Yes").astype(int)
)

# Display the new features
new_features = [
    "total_prior_visits",
    "medication_count",
    "medication_change_flag",
    "diabetes_medication_flag"
]
df_work = df_work.drop('weight', axis=1)

display(df_work[new_features].head())

print("New feature count:", len(new_features))
print("Updated dataset shape:", df_work.shape)
```
